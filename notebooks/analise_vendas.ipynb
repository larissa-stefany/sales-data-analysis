# Análise de Vendas
Notebook simples para explorar o dataset sintético de vendas e responder às principais perguntas de negócio do projeto.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'vendas.csv', parse_dates=['data'])
df['faturamento'] = df['quantidade'] * df['preco_unitario']
df.head()
```

```python
resumo = {
    'faturamento_total': df['faturamento'].sum(),
    'ticket_medio': df['faturamento'].mean(),
    'pedidos': df['pedido_id'].nunique(),
    'clientes': df['cliente_id'].nunique(),
}
pd.Series(resumo)
```

```python
categoria = df.groupby('categoria', as_index=False)['faturamento'].sum().sort_values('faturamento', ascending=False)
categoria
```

```python
produto = df.groupby('produto', as_index=False).agg(
    unidades=('quantidade', 'sum'),
    faturamento=('faturamento', 'sum')
).sort_values('faturamento', ascending=False)
produto.head(10)
```

```python
mensal = (
    df.assign(mes=df['data'].dt.to_period('M').astype(str))
      .groupby('mes', as_index=False)['faturamento'].sum()
)
ax = mensal.plot(x='mes', y='faturamento', marker='o', legend=False, figsize=(8,4))
ax.set_title('Evolução mensal do faturamento')
ax.set_xlabel('Mês')
ax.set_ylabel('Faturamento (R$)')
plt.xticks(rotation=45)
plt.tight_layout()
```

## Conclusão
A análise permite comparar faturamento por categoria, produto, cidade e mês. O foco do projeto é demonstrar um fluxo enxuto de análise de dados com Python e SQL, usando dados sintéticos e reproduzíveis.
